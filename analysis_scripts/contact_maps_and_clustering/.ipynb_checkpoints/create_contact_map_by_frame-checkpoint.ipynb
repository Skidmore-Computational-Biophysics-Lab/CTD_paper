# Create Contact Map Frame .csv Files 
## Purpose
This notebook is for creating .csv files comprised of the contact map frames from each construct. 

## Inputs
- .mdcrd files for all sets of simulations
- gas .prmtop files used for the simulations

## What it does
- Loads in data to create contact matrices.
- Turns each matrix into a binary frame of 1s (contact) and 0s (no contact).
- Appends all frames to make .csv files for each construct.
  
## Outputs
.csv files with contact map frames for each construct.

## Notes / Assumptions
This notebook assumes you have your .mdcrd files for all simulations.


```python
import mdtraj as md
import numpy as np
import os
import pandas as pd
from contact_map import ContactFrequency
from contact_map import ContactMap, ContactCount, ContactFrequency, ContactDifference, AtomMismatchedContactDifference
from numpy import *
import time
```

```python
start_time = time.time()
```

```python
asn_framenum = -4900000
con_framenum = -4800000
stride = 100000 #number of frames skipped - change to 100 after testing
k = 10
```

```python
#ASN VARIANT
traj_list_pa_cis = ["ps5n7_cis_combined_sim1.mdcrd", #simulations that start in cis
             "ps5n7_cis_combined_sim2.mdcrd", 
             "ps5n7_cis_combined_sim3.mdcrd", 
             "ps5n7_cis_combined_sim4.mdcrd", 
             "ps5n7_cis_combined_sim5.mdcrd"]

traj_list_pa_trans = ["ps5n7_trans_p_combined_sim1.mdcrd", #simulations that start in trans
             "ps5n7_trans_p_combined_sim2.mdcrd", 
             "ps5n7_trans_p_combined_sim3.mdcrd", 
             "ps5n7_trans_p_combined_sim4.mdcrd", 
             "ps5n7_trans_p_combined_sim5.mdcrd"]

#ASN VARIANT (UNphos)
traj_list_a_cis = ["s5n7_cis_combined_sim1.mdcrd",
             "s5n7_cis_combined_sim2.mdcrd", 
             "s5n7_cis_combined_sim3.mdcrd", 
             "s5n7_cis_combined_sim4.mdcrd", 
             "s5n7_cis_combined_sim5.mdcrd"]

traj_list_a_trans = ["s5n7_trans_combined_sim1.mdcrd",
             "s5n7_trans_combined_sim2.mdcrd", 
             "s5n7_trans_combined_sim3.mdcrd", 
             "s5n7_trans_combined_sim4.mdcrd", 
             "s5n7_trans_combined_sim5.mdcrd"]
```

```python
#CON VARIANT
traj_list_pc_cis = ["ps5s7_cis_p_combined_sim1.mdcrd", #simulations that start in cis
             "ps5s7_cis_p_combined_sim2.mdcrd", 
             "ps5s7_cis_p_combined_sim3.mdcrd", 
             "ps5s7_cis_p_combined_sim4.mdcrd", 
             "ps5s7_cis_p_combined_sim5.mdcrd"]

traj_list_pc_trans = ["ps5s7_trans_p_combined_sim1.mdcrd", #simulations that start in trans
             "ps5s7_trans_p_combined_sim2.mdcrd", 
             "ps5s7_trans_p_combined_sim3.mdcrd", 
             "ps5s7_trans_p_combined_sim4.mdcrd", 
             "ps5s7_trans_p_combined_sim5.mdcrd"]

#CON VARIANT (UNphos)
traj_list_c_cis = ["s5s7_cis_combined_sim1.mdcrd",
             "s5s7_cis_combined_sim2.mdcrd", 
             "s5s7_cis_combined_sim3.mdcrd", 
             "s5s7_cis_combined_sim4.mdcrd", 
             "s5s7_cis_combined_sim5.mdcrd"]

traj_list_c_trans = ["s5s7_trans_combined_sim1.mdcrd",
             "s5s7_trans_combined_sim2.mdcrd", 
             "s5s7_trans_combined_sim3.mdcrd", 
             "s5s7_trans_combined_sim4.mdcrd", 
             "s5s7_trans_combined_sim5.mdcrd"]
```

```python
#[using other pathway cell]
traj_new_pa = []
for suffix in traj_list_pa_cis:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/ps5n7/mdcrd/" + suffix,
                                       top="/data/chem_shared/CTD/ps5n7/ctd_cis_gas.prmtop", stride=stride)
    traj_new_pa.append(fulltraj[int(asn_framenum/stride):])

    print(suffix + " done")
    
for suffix in traj_list_pa_trans:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/ps5n7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/ps5n7/ctd_trans_p_gas.prmtop", stride=stride)
    traj_new_pa.append(fulltraj[int(asn_framenum/stride):])
    
    print(suffix + " done")

traj_new_a = []

for suffix in traj_list_a_cis:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/s5n7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/s5n7/ctd_cis_gas.prmtop", stride=stride)
    traj_new_a.append(fulltraj[int(asn_framenum/stride):])
    
    print(suffix + " done")
    
for suffix in traj_list_a_trans:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/s5n7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/s5n7/ctd_trans_gas.prmtop", stride=stride)
    traj_new_a.append(fulltraj[int(asn_framenum/stride):])
    
    print(suffix + " done")
```

```python
#[using other pathway cell]
traj_new_pc = []

for suffix in traj_list_pc_cis:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/ps5s7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/ps5s7/ps5s7_cis_gas.prmtop", stride=stride)
    traj_new_pc.append(fulltraj[int(con_framenum/stride):])
    
    print(suffix + " done")
    
for suffix in traj_list_pc_trans:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/ps5s7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/ps5s7/ps5s7_trans_gas.prmtop", stride=stride)
    traj_new_pc.append(fulltraj[int(con_framenum/stride):])
    
    print(suffix + " done")

#UNphos CON
traj_new_c = []

for suffix in traj_list_c_cis:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/s5s7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/s5s7/s5s7_cis_gas.prmtop", stride=stride)
    traj_new_c.append(fulltraj[int(con_framenum/stride):])
    
    print(suffix + " done")
    
for suffix in traj_list_c_trans:
    fulltraj = md.load_netcdf("/data/chem_shared/CTD/s5s7/mdcrd/" + suffix, 
                              top="/data/chem_shared/CTD/s5s7/s5s7_trans_gas.prmtop", stride=stride)
    traj_new_c.append(fulltraj[int(con_framenum/stride):])

    print(suffix + " done")
```

```python
topology_pa = traj_new_pa[0].topology
topology_a = traj_new_a[0].topology
topology_pc = traj_new_pc[0].topology
topology_c = traj_new_c[0].topology


# In[9]:


traj_combined_pa = md.join(traj_new_pa)
traj_combined_a = md.join(traj_new_a)
traj_combined_pc = md.join(traj_new_pc)
traj_combined_c = md.join(traj_new_c)


# In[10]:


traj_pa_combined = ContactFrequency(traj_combined_pa)
traj_a_combined = ContactFrequency(traj_combined_a)
traj_pc_combined = ContactFrequency(traj_combined_pc)
traj_c_combined = ContactFrequency(traj_combined_c)
```

```python
# Empty list to store the contact matrices for all frames
pa_contact_matrices = []
a_contact_matrices = []
pc_contact_matrices = []
c_contact_matrices = []

# Loop over each frame in the trajectory
for i in range(traj_combined_pa.n_frames):
    # Select the individual frame
    frame_traj_pa = traj_combined_pa[i]
    # Compute the contact map for the single frame
    frame_contacts_pa = ContactFrequency(frame_traj_pa)
    # Append the contact matrix (dataframe) to the list, replace NaNs with 0s
    pa_contact_matrices.append(frame_contacts_pa.residue_contacts.df.fillna(0))

for i in range(traj_combined_a.n_frames):
    frame_traj_a = traj_combined_a[i]
    frame_contacts_a = ContactFrequency(frame_traj_a)
    a_contact_matrices.append(frame_contacts_a.residue_contacts.df.fillna(0))

for i in range(traj_combined_pc.n_frames):
    frame_traj_pc = traj_combined_pc[i]
    frame_contacts_pc = ContactFrequency(frame_traj_pc)
    pc_contact_matrices.append(frame_contacts_pc.residue_contacts.df.fillna(0))

for i in range(traj_combined_c.n_frames):
    frame_traj_c = traj_combined_c[i]
    frame_contacts_c = ContactFrequency(frame_traj_c)
    c_contact_matrices.append(frame_contacts_c.residue_contacts.df.fillna(0))
```

```python
if os.path.exists("pa_contact_matrices_every_" + str(stride) + ".csv"):
  os.remove("pa_contact_matrices_every_" +  str(stride) + ".csv")
else:
  print("pa_contact_matrices_every_" +  str(stride) + ".csv does not exist")
for matrix in pa_contact_matrices:
    matrix.to_csv("pa_contact_matrices_every_" +  str(stride) + ".csv", mode='a')
```

```python
if os.path.exists("a_contact_matrices_every_" +  str(stride) + ".csv"):
  os.remove("a_contact_matrices_every_" +  str(stride) + ".csv")
else:
  print("a_contact_matrices_every_" +  str(stride) + ".csv does not exist")
for matrix in a_contact_matrices:
    matrix.to_csv("a_contact_matrices_every_" +  str(stride) + ".csv", mode='a')
```

```python
if os.path.exists("pc_contact_matrices_every_" +  str(stride) + ".csv"):
  os.remove("pc_contact_matrices_every_" +  str(stride) + ".csv")
else:
  print("pc_contact_matrices_every_" +  str(stride) + ".csv does not exist")
for matrix in pc_contact_matrices:
    matrix.to_csv("pc_contact_matrices_every_" +  str(stride) + ".csv", mode='a')
```

```python
if os.path.exists("c_contact_matrices_every_" +  str(stride) + ".csv"):
  os.remove("c_contact_matrices_every_" +  str(stride) + ".csv")
else:
  print("c_contact_matrices_every_" +  str(stride) + ".csv does not exist")
for matrix in c_contact_matrices:
    matrix.to_csv("c_contact_matrices_every_" +  str(stride) + ".csv", mode='a')
```
